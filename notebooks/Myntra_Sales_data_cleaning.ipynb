This notebook focuses on cleaning and preparing Myntra sales data for analysis and visualization.

- Data cleaning
- Data validation
- Feature engineering

```python
df = pd.read_csv("Myntra Sales Analysis.csv")
df.head()
```

```python
df = df.loc[:, ~df.columns.str.contains('^Unnamed', case = False)]
df = df.drop(columns = ['index'])
```

Removed unnecessary columns generated during CSV import (e.g., 'Unnamed' columns) to clean the dataset.

```python
df.head()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('-', '_')
```

```python
df['date'] = pd.to_datetime(df['date'], errors = 'coerce')
```

```python
df['ship_postal_code'] = df['ship_postal_code'].astype('Int64')
df['courier_status'] = df['courier_status'].fillna('Unkown')
```

```python
df.isnull().sum()
```

```python
df = df[df['amount'].notnull()].copy()
```

```python
df['currency'] = df['currency'].fillna('INR')
```

```python
df['promotion_ids'] = df['promotion_ids'].fillna('No Promotion')
df['fulfilled_by'] = df['fulfilled_by'].fillna('Unknown')
```

```python
df['revenue'] = df['qty'] * df['amount']
```

```python
df['month'] = df['date'].dt.month
```

Handled SettingWithCopyWarning by creating an explicit copy of the filtered DataFrame to ensure safe modifications.

```python
df.describe()
```

## Data Cleaning Steps

- Removed unnecessary columns ('index', unnamed columns)
- Standardized column names
- Converted date column to datetime format
- Handled missing values in key columns
- Created new features (revenue, month)

```python
df = df[df['qty'] > 0]
```

```python
df = df[df['amount'] > 0]
```

```python
df = df[df['revenue'] < 10000]
```

```python
df = df[df['ship_postal_code'].notnull()]
```
